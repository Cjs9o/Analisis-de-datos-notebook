## Ejercicio 8. Listas

Dada la lista:

```python

```

Calcule:

1. La cantidad de notas.
2. La nota máxima.
3. La nota mínima.
4. El promedio.
5. Las notas aprobadas.

````python
# Escriba su solución aquí
notas = {4.2, 3.5, 2.8, 4.7, 3.9, 2.5, 4.0}

cantidad = len(notas)
maxi = max(notas)
mini = min(notas)
prom = sum(notas)/len(notas)
aprobadas = []
for n in notas:
    if n >= 3.0:
        aprobadas.append(n)
print(f"Cantidad: {cantidad}, Maximo: {maxi}, Minimo: {mini},")
print(f"Promedio: {prom}, Aprovadas: {aprobadas}")
````

## Ejercicio 29. Aplicar una función

Cree una función que clasifique la satisfacción:

- Menor de 4.0: `Baja`
- Entre 4.0 y 4.4: `Media`
- 4.5 o más: `Alta`

Aplique la función y cree la columna `nivel_satisfaccion`.

````python
# Escriba su solución aquí
def clasificar_satisfaccion(valor):
    if valor < 4.0:
        return "Baja"
    elif valor < 4.5:
        return "Media"
    else:
        return "Alta"

datos_ventas["nivel_satisfaccion"] = datos_ventas["satisfaccion"].apply(clasificar_satisfaccion)
new_dv = datos_ventas[["vendedor", "satisfaccion", "nivel_satisfaccion"]]
display(new_dv)
````
